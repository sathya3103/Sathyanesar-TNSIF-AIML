```python
arr = [10, 5, 0, 7, 8, 0, 13, 4]

even_count = 0;
odd_count = 0;
zero_count = 0;

for i in range(len(arr)):

  if(arr[i] == 0):
    zero_count += 1


  elif(arr[i] % 2 == 0):
    even_count = even_count + 1


  else:
    odd_count = odd_count+1


print("Even :", even_count)
print("Odd :",odd_count)
print("Zero :",zero_count)
```

```python
arr = [1, 2, 3, 5, 6, 7]

n = max(arr)

expected_total = 0
actual_total = 0

for i in range(1, n + 1):
    expected_total = expected_total + i

for i in range(len(arr)):
    actual_total = actual_total + arr[i]

missing = expected_total - actual_total

print("Missing Number:", missing)
```

```python
arr = [2, 5, 2, 8, 5, 2, 3, 5, 2]

max_count = 0
max_element = 0

for i in range(len(arr)):
    count = 0

    for j in range(len(arr)):
        if arr[i] == arr[j]:
            count += 1

    if count > max_count:
        max_count = count
        max_element = arr[i]

print("Most Frequent Element:", max_element)
print("Frequency:", max_count)
```

```python
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

max_sum = arr[0]
max_subarray = []

for i in range(len(arr)):
    current_sum = 0

    for j in range(i, len(arr)):
        current_sum = current_sum + arr[j]

        if current_sum > max_sum:
            max_sum = current_sum
            max_subarray = arr[i:j+1]

print("Maximum Subarray Sum:", max_sum)
print("Subarray:", max_subarray)
```
